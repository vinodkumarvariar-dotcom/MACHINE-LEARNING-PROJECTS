● Provide a brief description of how KMeans clustering works

K-Means clustering is an unsupervised machine learning algorithm used to partition a dataset into a pre-specified number of distinct, non-overlapping subgroups, or "clusters". The algorithm works iteratively through the following steps

Initialization: The user defines the number of clusters, \(k\). The algorithm then randomly selects \(k\) data points as the initial cluster centers, also known as centroids

Assignment (Expectation Step): Each data point in the dataset is assigned to the nearest centroid, typically measured using squared Euclidean distance. This forms the initial clusters.

Update (Maximization Step): The centroids are recalculated by computing the mean of all data points assigned to that specific cluster. The center of the cluster moves to this new average position.

Convergence: Steps 2 and 3 are repeated until the centroids no longer move significantly or change their position between iterations, or a predefined number of iterations is reached. This indicates the clusters have stabilized

The objective of the K-Means algorithm is to minimize the within-cluster sum of squares (WCSS), which is the sum of the squared distances between each data point and the centroid of the cluster to which it belongs. This process results in compact and well-separated clusters

● Explain why KMeans clustering might be suitable for the Iris dataset

K-means clustering may be suitable for the Iris dataset because the data has characteristics that align well with the assumptions and goals of the K-means algorithm

The key reasons are

Known Number of Clusters: The Iris dataset is a well-known benchmark dataset that naturally contains exactly three distinct species (classes) of iris flowers: Setosa, Versicolor, and Virginica [1, 2]. K-means is a supervised algorithm in the sense that you must specify the number of clusters a priori (the "K" value). Since we know there are three species, we can set K=3, making it a perfect fit for the problem

Distinct and Separable Clusters: The original creators of the dataset structured it so that the three species generally form distinct, relatively compact clusters in the four-dimensional feature space (sepal length/width, petal length/width) [2]. K-means performs best when clusters are spherical, non-overlapping, and well-separated from one another, which is largely true for the Iris data

Numerical and Continuous Features: All features in the Iris dataset are quantitative and continuous, which is the type of data K-means is designed to handle, as it relies on calculating Euclidean distances between data points and cluster centroids

In summary, the Iris dataset's inherent structure—three well-defined, separable clusters with numerical features—makes it an excellent candidate for demonstrating and applying the K-means algorithm effectively

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

print("Keys of the iris data dictionary:", iris.keys())
print(f"Feature names: {iris.feature_names}")
print(f"Target names: {iris.target_names}")

X =  iris.data
y =  iris.target

df = pd.DataFrame(X, columns= iris.feature_names)

print("iris dataset Successfully Loaded")
print(df)
df.info()

import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Load and Preprocess the Iris dataset
# Load the dataset
iris = datasets.load_iris()
X = iris.data # Features: sepal length, sepal width, petal length, petal width
y = iris.target # True labels (for comparison)

# Standardize the features (good practice for both KMeans and PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Apply K-Means Clustering
# We know the Iris dataset has 3 species, so we set n_clusters=3
k = 3
kmeans = KMeans(n_clusters=k, init='k-means++', max_iter=300, n_init=10, random_state=0)
# Fit the model and predict the clusters for each data point
y_kmeans = kmeans.fit_predict(X_scaled)
# Get the cluster centers (centroids)
centroids = kmeans.cluster_centers_

# 3. Apply PCA for Visualization
# Reduce the data from 4 dimensions to 2 principal components for 2D plotting
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Transform the centroids to the 2D PCA space as well
centroids_pca = pca.transform(centroids)

# 4. Visualize the Clusters
plt.figure(figsize=(10, 8))

# Plot the data points, colored by their predicted cluster
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='viridis', s=50, label='Data points')

# Plot the centroids
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1], s=200, color='red', marker='X', label='Centroids')

plt.title('K-Means Clustering Visualization after PCA (Iris Dataset)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.colorbar(scatter, label='Cluster Label')
plt.grid(True)
plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

print("Keys of the iris data dictionary:", iris.keys())
print(f"Feature names: {iris.feature_names}")
print(f"Target names: {iris.target_names}")

X =  iris.data
y =  iris.target

df = pd.DataFrame(X, columns= iris.feature_names)

print("iris dataset Successfully Loaded")
print(df)
df.info()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Load and Preprocess the Iris dataset
iris = datasets.load_iris()
# Select the four features (sepal length, sepal width, petal length, petal width)
# and convert to a NumPy array for clustering. The 'species' column (target) is excluded.
X = iris.data[:, :4]

# It is good practice to scale the features for K-Means, as it is sensitive to the scale of the data.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Determine the optimal number of clusters (K) using the Elbow Method
# (Optional, but good practice. We expect K=3 for the Iris dataset).
wcss = [] # Within-cluster sum of squares
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_) # Inertia is the WCSS value

# Plotting the elbow method graph to observe the optimal K
plt.figure(figsize=(8, 4))
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters (K)')
plt.ylabel('WCSS')
plt.show()

# The elbow typically occurs at K=3, which is the optimal number of clusters for this dataset.

# 3. Apply K-Means clustering with K=3
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=0)
y_kmeans = kmeans.fit_predict(X_scaled) # Get the cluster labels for each data point

# 4. Visualize the clusters
# For visualization, we select two prominent features (e.g., Petal Length and Petal Width)
# as it's difficult to visualize all four dimensions effectively in a 2D plot.
plt.figure(figsize=(10, 7))
# Scatter plot for each cluster
plt.scatter(X_scaled[y_kmeans == 0, 2], X_scaled[y_kmeans == 0, 3], s = 100, c = 'red', label = 'Iris-setosa')
plt.scatter(X_scaled[y_kmeans == 1, 2], X_scaled[y_kmeans == 1, 3], s = 100, c = 'blue', label = 'Iris-versicolor')
plt.scatter(X_scaled[y_kmeans == 2, 2], X_scaled[y_kmeans == 2, 3], s = 100, c = 'green', label = 'Iris-virginica')

# Plot the centroids of the clusters
plt.scatter(kmeans.cluster_centers_[:, 2], kmeans.cluster_centers_[:, 3], s = 100, c = 'yellow', label = 'Centroids', marker='x')
plt.title('K-Means Clustering on Iris Dataset (Petal Length vs Petal Width)')
plt.xlabel('Petal Length (scaled)')
plt.ylabel('Petal Width (scaled)')
plt.legend()
plt.show()
```

● Provide a brief description of how Hierarchical clustering works.

Hierarchical clustering is an unsupervised machine learning algorithm that groups similar data points into a multi-level tree structure without requiring the number of clusters to be specified in advance. It works by measuring the similarity (or distance) between objects and organizing them into a hierarchy, typically visualized as a dendrogram
Core Approaches - The algorithm operates in one of two ways:

Agglomerative (Bottom-Up): The most common method, where every data point begins as its own individual cluster. The algorithm iteratively identifies and merges the two most similar clusters until only one large cluster remains.

Divisive (Top-Down): The entire dataset starts as a single cluster, which is then recursively split into smaller, more specific sub-clusters until each data point is isolated into its own singleton group.

How the Algorithm Works (Agglomerative)

1. Initial Calculation: A proximity matrix is created by calculating the distance between every pair of data points using a metric such as Euclidean distance.

2. Merging: The two clusters with the shortest distance (highest similarity) are merged into a single new cluster.

3. Linkage Update: The distance matrix is updated to reflect the distances between the newly formed cluster and all remaining clusters using a linkage method

Single Linkage: Uses the distance between the closest members of two clusters.

Complete Linkage: Uses the distance between the farthest members.

Average Linkage: Calculates the average of all pairwise distances between clusters.

Ward’s Method: Merges clusters that minimize the increase in total within-cluster variance

4. Iteration: Steps 2 and 3 are repeated until all points are contained within a single cluster.

Interpreting Results

The final output is a dendrogram, a tree-like diagram where the height of each merge indicates the distance or dissimilarity between the clusters. To determine the final number of clusters, an analyst "cuts" the dendrogram horizontally at a chosen height; the number of vertical lines intersected by this cut represents the final cluster count.

● Explain why Hierarchical clustering might be suitable for the Iris dataset.

Hierarchical clustering is highly suitable for the Iris dataset because it reflects the natural biological relationships and taxonomical structure of the flower species. As of 2026, it remains a standard method for exploring the dataset's underlying patterns.

Key reasons for its suitability include

1. Natural Biological Hierarchy

Taxonomical Relationships: The Iris dataset inherently represents biological categories (Setosa, Versicolor, and Virginica). Hierarchical clustering builds a "family tree" (dendrogram) that captures these nested relationships, showing how species are related at different levels of similarity

Species Differentiation: It excels at revealing that Iris setosa is a distinct, well-separated group, while Iris versicolor and Iris virginica share a closer hierarchical relationship and partial overlap in features like petal length and width.

2. No Need for Predefined Cluster Counts

Discovery-Oriented: Unlike K-Means, which requires you to specify the number of clusters (\(k=3\)) upfront, hierarchical clustering allows you to explore the data without this assumption

Optimal Cluster Identification: By examining the dendrogram, you can visually determine the most natural number of clusters by finding the largest gaps in vertical merge distances

3. Exploratory Power through Dendrograms

Multi-Level Visualization: The dendrogram provides a comprehensive overview of how every individual data point merges into groups, which is useful for identifying specific outliers or small sub-groups within the 150 samples

Customizable Cutting: You can "cut" the hierarchy at different levels to see how the data partitions into two, three, or more clusters, providing flexibility in how the species are interpreted

4. Data Compatibility

Minimal Outliers: The Iris dataset is relatively clean with few outliers, making it an ideal candidate for hierarchical methods which can otherwise be sensitive to noise

Small Dataset Size: With only 150 samples and 4 features, the computational cost of hierarchical clustering (\(O(N^{2})\) or \(O(N^{3})\)) is negligible, allowing for rapid and accurate processing

Silhout Score Comparison in Iris Data Set

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

print("Keys of the iris data dictionary:", iris.keys())
print(f"Feature names: {iris.feature_names}")
print(f"Target names: {iris.target_names}")

X =  iris.data
y =  iris.target

df = pd.DataFrame(X, columns= iris.feature_names)

print("iris dataset Successfully Loaded")
print(df)
df.info()

import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = load_iris()
X = iris.data

# Range of k values to evaluate (from 2 to 10 clusters)
# A minimum of 2 clusters is required to calculate the silhouette score
k_range = range(2, 11)
silhouette_scores = []

# Calculate silhouette score for each k
for k in k_range:
    # Initialize and fit the KMeans model
    # Using 'k-means++' initialization and n_init=10 for robust results
    kmeans_model = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    kmeans_model.fit(X)

    # Get the cluster labels
    cluster_labels = kmeans_model.labels_

    # Calculate the silhouette score and append to the list
    score = silhouette_score(X, cluster_labels)
    silhouette_scores.append(score)
    print(f"For n_clusters = {k}, the average silhouette score is: {score:.4f}")

# Plotting the silhouette scores
plt.figure(figsize=(8, 5))
plt.plot(k_range, silhouette_scores, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Average Silhouette Score')
plt.title('Silhouette Score Comparison for different k on Iris Dataset')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```
