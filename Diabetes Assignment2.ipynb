A. Central Tendency (Mean / Median / Mode) Find the mean of the Glucose values. Find the median of the BMI column. Find the mode of the BloodPressure column.

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")

import statistics
data = df["Glucose"]
mean_value = statistics.mean(data)
print("The mean is:", mean_value)
```

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
import numpy as np
data = df["BMI"]
median_np = np.median(data)
print(f"The median using NumPy is: {median_np}")
```

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
data = df["BloodPressure"]
series = pd.Series(data)

# Calling .mode() on a Series returns a Series, which can be converted to a list
modes_series = series.mode().tolist()
print(f"The modes are: {modes_series}")
```

B. Variance & Standard Deviation Find the variance of the Age column. Find the standard deviation of the BMI values. Find the variance of the Glucose column.

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
data = df["Age"]

# Sample Variance (ddof=1 is the default for pandas)
sample_variance = pd.Series(data).var()
print(f"Sample Variance: {sample_variance}")

# Population Variance (use the ddof=0 argument)
population_variance = pd.Series(data).var(ddof=0)
print(f"Population Variance: {population_variance}")
```

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
data = df["Glucose"]

# Sample Variance (ddof=1 is the default for pandas)
sample_variance = pd.Series(data).var()
print(f"Sample Variance: {sample_variance}")

# Population Variance (use the ddof=0 argument)
population_variance = pd.Series(data).var(ddof=0)
print(f"Population Variance: {population_variance}")
```

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
data = df["BMI"]
import numpy as np
# Population standard deviation (default behavior of numpy.std)
population_std_dev_np = np.std(data)
print(f"Numpy Population Std Dev: {population_std_dev_np}")

# Sample standard deviation (specify ddof=1 for Delta Degrees of Freedom)
sample_std_dev_np = np.std(data, ddof=1)
print(f"Numpy Sample Standard Deviation: {sample_std_dev_np}")
```

C. Skewness Find the skewness of the BMI column. Find the skewness of the Age column. Find the skewness of the Glucose values.

D. Kurtosis Find the kurtosis of the Glucose column. Find the kurtosis of the BMI column. Find the kurtosis of the BloodPressure column

```python
import pandas as pd
df = pd.read_excel("diabetes.csv.xlsx")
import numpy as np
from scipy.stats import kurtosis
data = df["Glucose"]
kurt_value = kurtosis(data)
print(f"Kurtosis (excess): {kurt_value}")
```
