```python
import pandas as pd
import numpy as np
df = pd.read_excel("ASSIGNMENT2.xlsx")
df['Height'] = pd.to_numeric(df['Height'], errors='coerce')
nan_height_mask = df['Height'].isnull()
df.loc[nan_height_mask, 'Height'] = np.random.randint(150, 181, size=nan_height_mask.sum())
df['Height'] = df['Height'].astype(int)
df[(df["Height"] > 150) & (df["Height"] < 181)]
```
