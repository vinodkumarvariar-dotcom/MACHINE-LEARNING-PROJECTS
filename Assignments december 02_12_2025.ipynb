How to compute the mean, median, and standard deviation of a numpy array?Use the
array you created above (Score 1)

```python
import numpy as np
my_array = np.array([1, 5, 2, 8, 3, 9, 4, 7, 6,7,8,9,10])
mean_value = np.mean(my_array)
print(f"Mean: {mean_value}")
```

```python
import numpy as np
my_array = np.array([1, 5, 2, 8, 3, 9, 4, 7, 6,7,8,9,10])
median_value = np.median(my_array)
print(f"Median: {median_value}")
```

Write a NumPy program that creates a 2D array x of shape (3, 4) and a 1D array y of
shape (4,). Subtract y from each row of x using broadcasting. (Score: 2)  

```python
import numpy as np

x = np.array([[10, 20, 30, 40],
              [11, 21, 31, 41],
              [12, 22, 32, 42]])

y = np.array([1, 2, 3, 4])
result = x - y

print("Original 2D array x:")
print(x)
print("\nOriginal 1D array y:")
print(y)
print("\nResult of x - y (broadcasting):")
print(result)
```
