Load the dataset and handle any missing data by replacing null values with the mean value of the respective column. (Score 2)

```python
import pandas as pd
df = pd.read_excel("Wine_Quality_red.xlsx")
df
import numpy as np
df.info()
# Identify numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

# Fill missing values in each numerical column with its mean
for col in numerical_cols:
    if df[col].isnull().any(): # Check if the column actually has missing values
        mean_value = df[col].mean()
        df[col] = df[col].fillna(mean_value)

```

Perform eigen decomposition on the covariance matrix you computed in question 3. Identify and interpret the results:Identify the top 2 eigenvalues of the covariance matrix,Identify the corresponding eigenvectors.(Score 2)

```python
import pandas as pd
df = pd.read_excel("Wine_Quality_red.xlsx")
import numpy as np

selected_columns = ['alcohol', 'density']
X = df[selected_columns].values
cov_matrix = np.cov(X.T)

eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

sorted_index = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_index]
sorted_eigenvectors = eigenvectors[:, sorted_index]

# Identify the top 2 eigenvalues and eigenvectors
# Assuming the matrix is at least 2x2
top_2_eigenvalues = sorted_eigenvalues[:2]
top_2_eigenvectors = sorted_eigenvectors[:, :2]

print("Covariance Matrix from Question 3:")
print(cov_matrix)
print(
"Eigenvalues (sorted descending):", sorted_eigenvalues)
print("Eigenvectors (sorted to match eigenvalues, columns):\n", sorted_eigenvectors)
print("\nTop 2 Eigenvalues:", top_2_eigenvalues)
print("Corresponding Top 2 Eigenvectors:\n", top_2_eigenvectors)

```
